# Task 2: Which molecule is it?

You are given four anonymous XYZ files in `unknowns/` and a molecule library in `library`.

## Task

Find the library molecule that corresponds to each unknown molecule.

Here, two structures are the same molecule if you can translate and rotate one structure onto the other, allowing for a different atom order. Do not use mirror reflections: a reflected chiral molecule is a different molecule.

Your final answer should contain one match for each file:

- `unknown_1.xyz`
- `unknown_2.xyz`
- `unknown_3.xyz`
- `unknown_4.xyz`

```python
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


def read_xyz(path: Path):
    lines = path.read_text().strip().splitlines()
    n_atoms = int(lines[0])
    rows = [line.split() for line in lines[2 : 2 + n_atoms]]
    elements = [row[0] for row in rows]
    coords = np.array([[float(x), float(y), float(z)] for _, x, y, z in rows])
    return elements, coords


base_dir = Path('.')
if not (base_dir / 'unknowns').exists():
    base_dir = Path('task2')

unknown_files = sorted((base_dir / 'unknowns').glob('unknown_*.xyz'))
library_files = sorted((base_dir / 'library').glob('*.xyz'))

print(f'{len(unknown_files)} unknown files')
print(f'{len(library_files)} library files')
```

```python
rows = []
for path in unknown_files:
    elements, coords = read_xyz(path)
    rows.append({
        'file': path.name,
        'atoms': len(elements),
        'elements': ' '.join(sorted(set(elements))),
    })

pd.DataFrame(rows)
```
